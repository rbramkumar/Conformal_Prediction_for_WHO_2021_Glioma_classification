```python
# table1.py
# Table 1: the six rules at the 90% target, mean over the 200 test halves.
# Runs on its own. Reads core_outputs/decision_rules_200_splits.csv 
# and writes core_outputs/table1.csv.

import os
import pandas as pd

OUT = 'core_outputs'
rules = pd.read_csv(os.path.join(OUT, 'decision_rules_200_splits.csv'))

# rule type, rule name in the csv, rule name in the manuscript, in table order
rule_groups = [('Single-label rules', 'Recalibrated argmax', 'Recalibrated argmax'),
               ('Single-label rules', 'Locally weighted argmax', 'Class-weighted argmax'),
               ('Probability-based rules', 'Recalibrated probability sets', 'Probability sets'),
               ('Probability-based rules', 'Probability abstention', 'Probability-based abstention'),
               ('Conformal prediction', 'Single-threshold conformal', 'Standard conformal'),
               ('Conformal prediction', 'Mondrian conformal', 'Class-conditional conformal')]

means = rules.groupby(['cohort', 'rule']).mean(numeric_only=True)     # mean of every column over the 200 splits

table1_rows = []
for cohort in ['EGD', 'UTSW']:
    for group, rule, name in rule_groups:
        m = means.loc[(cohort, rule)]
        table1_rows.append({'Cohort': cohort, 'Rule type': group, 'Rule': name,
                            'Coverage, astrocytoma': round(m['cov_A'], 2),
                            'Coverage, glioblastoma': round(m['cov_G'], 2),
                            'Coverage, oligodendroglioma': round(m['cov_O'], 2),
                            'Mean set size': round(m['size'], 2),
                            'Singleton yield (%)': round(m['single_pct'], 1),
                            'Singleton accuracy (%)': round(m['acc_single_pct'], 1)})

table1 = pd.DataFrame(table1_rows)
table1.to_csv(os.path.join(OUT, 'table1.csv'), index=False)
print(table1)
```
