```python
# Supplemental figure: ROC and precision-recall curves of the calibrated random forest, one subtype against the rest.
# Columns: development (out-of-fold), EGD, UTSW-Glioma. Row 1 (A-C): ROC. Row 2 (D-F): precision-recall, dotted line = prevalence (AUPRC of a random classifier).
# Runs on its own. Reads modeling_table.csv and core_outputs/oof_probas.npy, egd_probas.npy, utsw_probas.npy (rev_1_1_code.ipynb).
# Writes core_outputs/FigureS_ROC_PR.png and .tiff

import os
import numpy as np
import pandas as pd
from sklearn.metrics import roc_curve, precision_recall_curve, roc_auc_score, average_precision_score
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

OUT = 'core_outputs'
table = pd.read_csv('modeling_table.csv', usecols=['site', '2021_glioma_class'])
classes = np.array(['astrocytoma', 'glioblastoma', 'oligodendroglioma'])
y_dev = table[table['site'].isin(['TCGA', 'UCSF'])]['2021_glioma_class'].values
y_egd = table[table['site'] == 'EGD']['2021_glioma_class'].values
y_utsw = table[table['site'] == 'UTSW']['2021_glioma_class'].values
oof = np.load(os.path.join(OUT, 'oof_probas.npy'))
egd = np.load(os.path.join(OUT, 'egd_probas.npy'))
utsw = np.load(os.path.join(OUT, 'utsw_probas.npy'))

plt.rcParams.update({'font.family': ['Arial', 'Liberation Sans', 'DejaVu Sans'], 'font.size': 8})
colors = ['#0072B2', '#D55E00', '#009E73']        # astrocytoma, glioblastoma, oligodendroglioma (same as Figure 2)
names = ['Astrocytoma', 'Glioblastoma', 'Oligodendroglioma']
cohorts = [('A', 'D', 'Development (out-of-fold)', y_dev, oof), ('B', 'E', 'EGD', y_egd, egd), ('C', 'F', 'UTSW-Glioma', y_utsw, utsw)]

fig, axes = plt.subplots(2, 3, figsize=(10, 7.4))

for col, (letter, letter_pr, title, y, p) in enumerate(cohorts):
    ax_roc = axes[0, col]
    ax_pr = axes[1, col]
    for k in range(3):
        truth = y == classes[k]
        fpr, tpr, _ = roc_curve(truth, p[:, k])
        precision, recall, _ = precision_recall_curve(truth, p[:, k])
        auc = roc_auc_score(truth, p[:, k])
        auprc = average_precision_score(truth, p[:, k])
        prevalence = truth.mean()
        ax_roc.plot(fpr, tpr, color=colors[k], linewidth=1.3, label='{} (AUC {:.2f})'.format(names[k], auc))
        ax_pr.step(recall, precision, where='post', color=colors[k], linewidth=1.3, label='{} (AUPRC {:.2f})'.format(names[k], auprc))
        ax_pr.axhline(prevalence, color=colors[k], linestyle=':', linewidth=0.9)
    ax_roc.plot([0, 1], [0, 1], color='gray', linestyle='--', linewidth=0.8)
    ax_roc.set_title(letter + '   ' + title + ': ROC', loc='left', fontweight='bold', fontsize=9)
    ax_roc.set_xlabel('1 - specificity')
    ax_roc.set_ylabel('Sensitivity')
    ax_roc.legend(loc='lower right', frameon=False, fontsize=7)
    ax_pr.set_title(letter_pr + '   ' + title + ': precision-recall', loc='left', fontweight='bold', fontsize=9)
    ax_pr.set_xlabel('Recall (sensitivity)')
    ax_pr.set_ylabel('Precision (PPV)')
    ax_pr.legend(loc='upper center', bbox_to_anchor=(0.5, -0.2), frameon=False, fontsize=7)
    ax_roc.set_xlim(0, 1)
    ax_roc.set_ylim(0, 1.02)
    ax_pr.set_xlim(0, 1)
    ax_pr.set_ylim(0, 1.02)
    ax_roc.grid(color='#eeeeee', linewidth=0.6)
    ax_pr.grid(color='#eeeeee', linewidth=0.6)
    ax_roc.spines['top'].set_visible(False)
    ax_roc.spines['right'].set_visible(False)
    ax_pr.spines['top'].set_visible(False)
    ax_pr.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig(os.path.join(OUT, 'FigureS_ROC_PR.png'), dpi=300)
plt.savefig(os.path.join(OUT, 'FigureS_ROC_PR.tiff'), dpi=600, pil_kwargs={'compression': 'tiff_lzw'})
```
