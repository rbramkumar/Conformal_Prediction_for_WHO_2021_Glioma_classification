```python
# figure2.py
# Figure 2: class-conditional (Mondrian) conformal prediction at the four targets, 200 splits per cohort.
# Runs on its own. Reads core_outputs/conformal_200_splits.csv (glioma_cp_core.py, section 4), writes core_outputs/Figure2.png/.tiff
# Rows: EGD, UTSW-Glioma. Columns: coverage, mean set size (empty sets count as 0).
# Point = mean over the 200 splits. Bar = 2.5th to 97.5th percentile across splits (split-to-split variability).

import os
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

OUT = 'core_outputs'
cp = pd.read_csv(os.path.join(OUT, 'conformal_200_splits.csv'))     # one row per cohort, split, target and subtype

plt.rcParams.update({'font.family': ['Arial', 'Liberation Sans', 'DejaVu Sans'], 'font.size': 8})
subtypes = [('astrocytoma', 'Astrocytoma', '#0072B2', 'o'),
            ('glioblastoma', 'Glioblastoma', '#D55E00', 's'),
            ('oligodendroglioma', 'Oligodendroglioma', '#009E73', '^')]
targets = [0.80, 0.85, 0.90, 0.95]

fig, axes = plt.subplots(2, 2, figsize=(6, 5))
panels = [('EGD', 'covered', 'A', 'EGD'), ('EGD', 'size', 'B', 'EGD'),
          ('UTSW', 'covered', 'C', 'UTSW-Glioma'), ('UTSW', 'size', 'D', 'UTSW-Glioma')]

for ax, (cohort, metric, letter, name) in zip(axes.ravel(), panels):
    for i, (cls, label, color, mark) in enumerate(subtypes):
        d = cp[(cp['cohort'] == cohort) & (cp['class'] == cls)]          # 200 splits x 4 targets for this subtype
        m = d.groupby('target')[metric].mean()
        lo = d.groupby('target')[metric].quantile(0.025)
        hi = d.groupby('target')[metric].quantile(0.975)
        x = m.index + (i - 1) * 0.008                                   # small offset so the bars don't overlap
        ax.errorbar(x, m, yerr=[m - lo, hi - m], color=color, marker=mark, markersize=4, linewidth=1.1,
                    capsize=2, elinewidth=0.8, label=label)
    if metric == 'covered':
        ax.plot(targets, targets, color='black', linestyle='--', linewidth=0.8, label='Target')
        ax.set_ylabel('Coverage')
        ax.set_ylim(0.5, 1.02)
    else:
        ax.set_ylabel('Mean set size (subtypes per patient)')
        ax.set_ylim(0.8, 3.0)
    ax.set_xticks(targets)
    ax.set_xticklabels(['80%', '85%', '90%', '95%'])
    ax.set_xlabel('Target')
    ax.set_title(letter + '   ' + name + (': coverage' if metric == 'covered' else ': set size'), loc='left', fontweight='bold', fontsize=9)
    ax.grid(color='#eeeeee', linewidth=0.6)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

fig.legend(*axes[0, 0].get_legend_handles_labels(), loc='lower center', ncol=4, frameon=False)
plt.tight_layout(rect=[0, 0.05, 1, 1])
plt.savefig(os.path.join(OUT, 'Figure2.png'), dpi=600)
plt.savefig(os.path.join(OUT, 'Figure2.tiff'), dpi=600, pil_kwargs={'compression': 'tiff_lzw'})

# share of splits below the target, per cohort, target and subtype (for the Results text)
cp['below'] = cp['covered'] < cp['target']
print((cp.groupby(['cohort', 'target', 'class'])['below'].mean() * 100).round(0).unstack())
```
